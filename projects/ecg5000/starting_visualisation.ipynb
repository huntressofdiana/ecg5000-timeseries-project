```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
train = pd.read_csv(
    "data/ECG5000_TRAIN.txt",
    sep=r"\s+",
    header=None
)

train.head()
```

```python
train.shape
```

```python
X_train = train.iloc[:, 1:]
y_train = train.iloc[:, 0]

print("ECG signals:", X_train.shape)
print("Labels:", y_train.shape)
```

```python
# Looking at just 1 heartbeat
plt.plot(X_train.iloc[0])

plt.xlabel("Time Step")
plt.ylabel("Amplitude")
plt.title(f"ECG Waveform - Class {int(y_train.iloc[0])}")

plt.show()
```

```python
classes = sorted(y_train.unique())

for c in classes:
    index = y_train[y_train == c].index[0]

    plt.plot(
        X_train.loc[index],
        label=f"Class {int(c)}"
    )

plt.xlabel("Time Step")
plt.ylabel("Amplitude")
plt.title("Example ECG Waveform from Each Class")
plt.legend()
plt.show()
```

```python
# Count how many ECGs belong to each class
class_counts = y_train.value_counts().sort_index()

print(class_counts) 

plt.bar(class_counts.index, class_counts.values)

plt.xlabel("Class")
plt.ylabel("Number of ECGs")
plt.title("Class Distribution in Training Data")
plt.xticks(class_counts.index)

plt.show()
```
