# Univariate and Multivariate Time Series

Showing some examples for univariate and multivariate time series data (real measurements)

## Physio Data

BIDMC PPG dataset from [**PhysioNet**](https://physionet.org/content/bidmc/1.0.0/)

contains 53 patient recordings, each eight minutes long at sampling rate 1Hz:
- Heart rate (HR)
- Respiratory rate (RR)
- Oxygen saturation (SpO₂)
- Pulse rate
- ECG, PPG, and respiration waveforms

Using [patient's 1 data](https://physionet.org/files/bidmc/1.0.0/bidmc_csv/bidmc_01_Numerics.csv)

```python
import pandas as pd
import matplotlib.pyplot as plt
url = ("https://physionet.org/files/bidmc/1.0.0/bidmc_csv/bidmc_01_Numerics.csv")
df = pd.read_csv(url)

time = df.index / 60  # observations are one second apart
df.columns = df.columns.str.strip() # strip whitespace from column names
print(repr(df.columns.tolist()))
```

### univariate sequence

let's pick the heart rate. single channel measured over time

```python
plt.figure(figsize=(10, 3))
plt.plot(time, df["HR"], color="black")
plt.xlabel("Time (minutes)")
plt.ylabel("Heart rate (bpm)")
plt.show()
```

### multivariate sequence

multiple channels / sensors recorded synchronously over time

```python
fig, axes = plt.subplots(3, 1, figsize=(10, 7), sharex=True)

for ax, variable, label in zip(
    axes,
    ["HR", "RESP", "SpO2"],
    ["Heart rate (bpm)", "Respiratory rate (breaths/min)", "SpO₂ (%)"]
):
    ax.plot(time, df[variable], color="black")
    ax.set_ylabel(label)

axes[-1].set_xlabel("Time (minutes)")
fig.suptitle("Multivariate patient monitoring")
plt.tight_layout()
plt.show()
```
